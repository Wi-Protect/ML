# Data Preprocessing

Using multiple antennas is an effective method to improve the detection performance, which will lead to a higher computation complex. Consequently, data fusion is the first step in feature extraction. <br>
The __median amplitudes__ of CSI are chosen as the final amplitudes, which are from the subcarriers of the same frequency.

```python
import numpy as np

# Amplitude data only.
X1 = df.iloc[:, 2:66]
X2 = df.iloc[:, 130:194]

# Ensure X1 and X2 have the same shape
assert X1.shape == X2.shape, "X1 and X2 must have the same shape"

# Initialize X with the same shape as X1 and X2
X = pd.DataFrame(index=X1.index, columns=X1.columns)

# Compute the median for each cell
for i in range(X1.shape[0]):
    for j in range(X1.shape[1]):
        X.iloc[i, j] = np.median([X1.iloc[i, j], X2.iloc[i, j]])

Y = df.iloc[:, -1]
```

```python
# Read from the the data to a CSV file to get rid of the data preprocessing step in the future.
data = pd.read_csv('median_amplitudes.csv')
X = data.iloc[:, :-1]
Y = data.iloc[:, -1]
```

# Feature extraction

```python
sample_rate = 25  # Hz
window = 1    # Window size in seconds
window_size = window * sample_rate

num_subcarriers = 64
```

```python
features = []
labels = []

for window_start in range(0, len(X) - window_size + 1 , window_size):
    window_end = window_start + window_size
    window_data = X.iloc[window_start:window_end]
    
    # Calculate the variance of the amplitudes of each subcarrier within the sliding window
    V_w = np.var(window_data, axis=0)
    # Calculate the variance of all subcarriers' variances
    V = np.var(V_w)
    features.append([V])

    labels.append(Y.iloc[window_end - 1])   # Assuming the label for the window is the label of the last sample in the window
```

```python
features = pd.DataFrame(features, columns=['V'])
```

# Classification Model

Split the data into training and validation sets

```python
from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(features_df, labels_df, test_size=0.2, random_state=42)
```

```python
from hmmlearn import hmm

"""
Train an HMM using the extracted features.
Args:
    features: Extracted features.
    n_states: Number of hidden states.
    n_segments: Number of observed states (segments).
Returns:
    model: Trained HMM model.
"""
n_states = 2
n_segments = 5

# Segment the features
thresholds = np.linspace(X_train['V'].min(), X_train['V'].max(), n_segments + 1)
observed_states = np.digitize(X_train, thresholds) - 1

# Train HMM
model = hmm.MultinomialHMM(n_components=n_states, n_iter=1000)
model.fit(observed_states.reshape(-1, 1))
```

### SVM Parameters
They adopted the radial basis function (RBF) kernel (a.k.a. Gaussian kernel) as the kernel function so that a non-linear regression model can be constructed. 
This SVM has two such training factors: C which controls overfitting of the model, and γ which controls the degree of nonlinearity of the model. In our system, we set C = 0.8 and γ = 20.



```python
from sklearn import svm

model = svm.SVC(C = 0.8 , gamma = 20 , kernel = 'rbf')

model.fit(X_train, y_train)
```
