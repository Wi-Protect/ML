# Down Sample the Data

choose data points from the data set to make 25 record for each second

```python
groupedData = df.groupby('time')
```

```python
selectedRows = []

newDataFrame = pd.DataFrame(columns=df.columns)
```

```python
groupedDataSizes = groupedData.size()
```

```python
print(groupedDataSizes)
```

```python
groupedDataSizes[0]
```

```python
cannotGetCount = 0
canGetCount = 0
```

```python
for i in range(len(groupedData)):
	count = groupedDataSizes.iloc[i]
	if count >= 25:
		canGetCount += 1
		fraction = count/ 25
		for j in range(1,26):
			index = int(j * fraction)
			group = groupedData.get_group(groupedDataSizes.index[i])
			if 0 <= index-1 < len(group):
				selectedRows.append(group.iloc[index-1])
			else:
				print(f"Index {index-1} out of bounds for group with {len(group)} rows")

		
		
	else:
		cannotGetCount += 1
		

```

```python
downsampledData = pd.DataFrame(selectedRows)
```
